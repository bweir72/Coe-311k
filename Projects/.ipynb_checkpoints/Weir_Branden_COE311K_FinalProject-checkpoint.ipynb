```python
# Convert second-order ODE to first-order system
# y1 = y
# y2 = dy/dt
def f(t, Y):
    y1, y2 = Y
    dy1dt = y2
    dy2dt = -(k/m) * y1 - (c/m) * y2
    return np.array([dy1dt, dy2dt])
```

```python
def euler_method(h, m, c, k, t_start, t_end, y_0, v_0):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    v_values = np.zeros(len(t_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(t_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-(c/m) * v_n - (k/m) * y_n)

    return t_values, y_values, v_values
```

```python
def rk4_method(h, m, c, k, t_start, t_end, y_0, v_0):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    v_values = np.zeros(len(t_values))

    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(t_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        def f1(y, v):
            return v

        def f2(y, v):
            return -(c/m) * v - (k/m) * y

        k1y = h * f1(y_n, v_n)
        k1v = h * f2(y_n, v_n)

        k2y = h * f1(y_n + 0.5 * k1y, v_n + 0.5 * k1v)
        k2v = h * f2(y_n + 0.5 * k1y, v_n + 0.5 * k1v)

        k3y = h * f1(y_n + 0.5 * k2y, v_n + 0.5 * k2v)
        k3v = h * f2(y_n + 0.5 * k2y, v_n + 0.5 * k2v)

        k4y = h * f1(y_n + k3y, v_n + k3v)
        k4v = h * f2(y_n + k3y, v_n + k3v)

        y_values[i] = y_n + (k1y + 2*k2y + 2*k3y + k4y) / 6
        v_values[i] = v_n + (k1v + 2*k2v + 2*k3v + k4v) / 6

    return t_values, y_values, v_values
```

```python
# Parameters for Concorde 102 main gear
m = 5.2e4       # kg
c = 3.5e5       # N*s/m
k = 2.5e6       # N/m

# Initial conditions
t0 = 0.0
tf = 5.0
y0 = 0.0        # initial displacement (m)
v0 = -3.05      # initial velocity (m/s)

# Initial state vector
Y0 = np.array([y0, v0])

# Example step sizes
h_values = [0.1, 0.05, 0.01]

for h in h_values:
    t_e, Y_e = euler_forward(f, t0, tf, Y0, h)
    t_r, Y_r = rk4(f, t0, tf, Y0, h)

    plt.figure()
    plt.plot(t_e, Y_e[:, 0], label=f"Euler, h={h}")
    plt.plot(t_r, Y_r[:, 0], label=f"RK4, h={h}")
    plt.xlabel("Time (s)")
    plt.ylabel("Displacement y(t) (m)")
    plt.title("Concorde 102 Main Gear Response")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters for Concorde 102 main gear
m = 5.2e4       # kg
c = 3.5e5       # N*s/m
k = 2.5e6       # N/m

# Initial conditions
t0 = 0.0
tf = 5.0
y0 = 0.0        # initial displacement (m)
v0 = -3.05      # initial velocity (m/s)

# Convert second-order ODE to first-order system
# y1 = y
# y2 = dy/dt
def f(t, Y):
    y1, y2 = Y
    dy1dt = y2
    dy2dt = -(k/m) * y1 - (c/m) * y2
    return np.array([dy1dt, dy2dt])

def euler_forward(f, t0, tf, Y0, h):
    t = np.arange(t0, tf + h, h)
    Y = np.zeros((len(t), len(Y0)))
    Y[0] = Y0

    for n in range(len(t) - 1):
        Y[n+1] = Y[n] + h * f(t[n], Y[n])

    return t, Y

def rk4(f, t0, tf, Y0, h):
    t = np.arange(t0, tf + h, h)
    Y = np.zeros((len(t), len(Y0)))
    Y[0] = Y0

    for n in range(len(t) - 1):
        k1 = f(t[n], Y[n])
        k2 = f(t[n] + h/2, Y[n] + (h/2) * k1)
        k3 = f(t[n] + h/2, Y[n] + (h/2) * k2)
        k4 = f(t[n] + h,   Y[n] + h * k3)

        Y[n+1] = Y[n] + (h/6) * (k1 + 2*k2 + 2*k3 + k4)

    return t, Y

# Initial state vector
Y0 = np.array([y0, v0])

# Example step sizes
h_values = [0.1, 0.05, 0.01]

for h in h_values:
    t_e, Y_e = euler_forward(f, t0, tf, Y0, h)
    t_r, Y_r = rk4(f, t0, tf, Y0, h)

    plt.figure()
    plt.plot(t_e, Y_e[:, 0], label=f"Euler, h={h}")
    plt.plot(t_r, Y_r[:, 0], label=f"RK4, h={h}")
    plt.xlabel("Time (s)")
    plt.ylabel("Displacement y(t) (m)")
    plt.title("Concorde 102 Main Gear Response")
    plt.legend()
    plt.show()
```
