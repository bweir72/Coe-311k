```python
import numpy as np
import matplotlib.pyplot as plt

# Convert second-order ODE to first-order system
# y1 = y
# y2 = dy/dt
def f(t, Y):
    y1, y2 = Y
    dy1dt = y2
    dy2dt = -(k/m) * y1 - (c/m) * y2
    return np.array([dy1dt, dy2dt])



def euler_method(h, m, c, k, t_start, t_end, y_0, v_0):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    v_values = np.zeros(len(t_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(t_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-(c/m) * v_n - (k/m) * y_n)

    return t_values, y_values, v_values




def rk4_method(h, m, c, k, t_start, t_end, y_0, v_0):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    v_values = np.zeros(len(t_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(t_values)):
        t_n = t_values[i - 1]
        Y_n = np.array([y_values[i - 1], v_values[i - 1]])

        k1 = h * f(t_n, Y_n)
        k2 = h * f(t_n + h/2, Y_n + k1/2)
        k3 = h * f(t_n + h/2, Y_n + k2/2)
        k4 = h * f(t_n + h, Y_n + k3)

        Y_next = Y_n + (k1 + 2*k2 + 2*k3 + k4) / 6

        y_values[i] = Y_next[0]
        v_values[i] = Y_next[1]

    return t_values, y_values, v_values



def exact_solution(t, m, c, k, v0):
    natural_freq  = np.sqrt(k / m)
    damping_ratio = c / (2 * np.sqrt(m * k))
    damped_freq   = natural_freq * np.sqrt(1 - damping_ratio**2)
    y_exact = (v0 / damped_freq) * np.exp(-damping_ratio * natural_freq * t) * np.sin(damped_freq * t)
    return y_exact



def Euler_Stability_Analysis(A, k, m, c, h_values):
    print(f"Euler stabilty")
    for h in h_values:
        # Build the amplification matrix G = I + h*A
        G = np.eye(2) + h * A
        
        # Compute eigenvalues of G for this step size
        G_eigenvalues,_  = np.linalg.eig(G)
        
        # Magnitudes must be <= 1 to be stable
        magnitudes = np.abs(G_eigenvalues)
        
        # Print step size and stability status
        print(f"\nStep size h = {h}")
        
        if np.all(magnitudes <= 1):
            print("Solution is Stable")
        else:
            print("Solution may not be Entirely Stable ;)")

    # Find maximum stable step size by testing increasing h values
    h_max_euler = 0

    for h_test in np.arange(0.01, 2.0, 0.001):
        G_test = np.eye(2) + h_test * A
        G_eigs, _ = np.linalg.eig(G_test)

        if any(np.abs(G_eigs) > 1):
            h_max_euler = h_test - 0.001
            break
    return h_max_euler



def rk4_stability_function(z):
        R = 1 + z + (z**2)/2 + (z**3)/6 + (z**4)/24
        return R
    
    
    
def RK4_Stability_Analysis(A, h_values):
    eigenvalues, _ = np.linalg.eig(A)
    
    print("\nRK4 Stability")
    
    # Checking stability for each h
    for h in h_values:
        stable = True
        for lam in eigenvalues:
            z = h * lam
            if abs(rk4_stability_function(z)) > 1:
                stable = False
                break
        
        print(f"\nStep size h = {h}")
        if stable:
            print("Solution is Stable")
        else:
            print("Solution may not be Entirely Stable ;)")
    
    # Find max stable step size
    h_max_rk4 = 0
    
    for h_test in np.arange(0.001, 2.0, 0.001):
        stable = True
        for lam in eigenvalues:
            z = h_test * lam
            if abs(rk4_stability_function(z)) > 1:
                stable = False
                break
        if not stable:
            h_max_rk4 = h_test - 0.001
            break
            
    return h_max_rk4



def calculate_error(t_values, y_numerical, m, c, k, v0):
    y_exact = exact_solution(t_values, m, c, k, v0)
    error = np.abs(y_numerical - y_exact)
    return error



# Parameters for Concorde 102 main gear
m = 5.2e4       # kg
c = 3.5e5       # N*s/m
k = 2.5e6       # N/m

# Initial conditions
t0 = 0.0
tf = 5.0
y0 = 0.0        # initial displacement (m)
v0 = -3.05      # initial velocity (m/s)

# Coefficient Matrix
A = np.array([[  0,    1],
                  [-k/m, -c/m]])
    
eigenvalues,_  = np.linalg.eig(A)

# Step sizes
h_values = [0.01, 0.03, 0.05, 0.07, 0.1]

# Analytical solution
t_exact = np.linspace(t0, tf, 1000)
y_exact = exact_solution(t_exact, m, c, k, v0)

# Euler Max step size
h_max_euler = Euler_Stability_Analysis(A, k, m, c, h_values)
print(f"\nMaximum stable step size for Euler: h = {h_max_euler}")

# Rk4 Max step size
h_max_rk4 = RK4_Stability_Analysis(A, h_values)
print(f"\nMaximum stable step size for RK4: h = {h_max_rk4:.4f} s\n")

# Unstable Analysis
h_unstable = .17
t_e_unstable, y_e_unstable, v_e_unstable = euler_method(h_unstable, m, c, k, t0, tf, y0, v0)

# Collect max error at each step size
euler_max_errors = []
rk4_max_errors = []

for h in h_values:
    t_e, y_e, v_e = euler_method(h, m, c, k, t0, tf, y0, v0)
    error_e = calculate_error(t_e, y_e, m, c, k, v0)
    euler_max_errors.append(np.max(error_e))
    
    t_r, y_r, v_r = rk4_method(h, m, c, k, t0, tf, y0, v0)
    error_r = calculate_error(t_r, y_r, m, c, k, v0)
    rk4_max_errors.append(np.max(error_r))

# Plotting Euler Displacement
plt.figure(figsize=(12, 6))
for h in h_values:
    t_e, y_e, v_e = euler_method(h, m, c, k, t0, tf, y0, v0)

    plt.plot(t_e, y_e, label=f"Euler, h={h}")
plt.plot(t_exact, y_exact, 'k--', label="Exact Solution")
plt.xlabel("Time (s)")
plt.ylabel("Displacement y(t) (m)")
plt.title("Displacement of Concorde 102 Main Gear")
plt.legend()
plt.grid(True)
plt.show()

# Plotting Euler at unstable h
plt.figure(figsize=(12, 6))
plt.plot(t_e_unstable, y_e_unstable, label=f"Euler, h={h_unstable}")
plt.xlabel("Time (s)")
plt.ylabel("Displacement y(t) (m)")
plt.title("Unstable Concorde 102 Main Gear Response")
plt.legend()
plt.grid(True)
plt.show()

# Plotting Euler Velocity
plt.figure(figsize=(12, 6))
for h in h_values:
    t_e, y_e, v_e = euler_method(h, m, c, k, t0, tf, y0, v0)

    plt.plot(t_e, v_e, label=f"Euler, h={h}")
plt.xlabel("Time (s)")
plt.ylabel("Velocity v(t) (m/s)")
plt.title("Velocity of Concorde 102 Main Gear Response")
plt.legend()
plt.grid(True)
plt.show()

# Phase portraits of Euler
plt.figure(figsize=(12, 6))
for h in h_values:
    t_e, y_e, v_e = euler_method(h, m, c, k, t0, tf, y0, v0)

    plt.plot(y_e, v_e, label=f"Euler, h={h}")
plt.xlabel("Displacement")
plt.ylabel("Velocity")
plt.title("Velocity vs Displacement using Euler's")
plt.legend()
plt.grid(True)
plt.show()

# Plot Euler error
plt.figure(figsize=(12, 6))

for h in h_values:
    t_e, y_e, v_e = euler_method(h, m, c, k, t0, tf, y0, v0)
    error_e = calculate_error(t_e, y_e, m, c, k, v0)
    
    plt.plot(t_e, error_e, label=f"Euler h={h}")
plt.xlabel("Time (s)")
plt.ylabel("Absolute Error (m)")
plt.title("Euler Method Error Over Time")
plt.legend()
plt.grid(True)
plt.show()

# Plotting RK4 Displacement
plt.figure(figsize=(12, 6))
for h in h_values:
    t_r, y_r, v_r = rk4_method(h, m, c, k, t0, tf, y0, v0)
    
    plt.plot(t_r, y_r, label=f"RK4, h={h}")
plt.plot(t_exact, y_exact, 'k--', label="Exact Solution")
plt.xlabel("Time (s)")
plt.ylabel("Displacement y(t) (m)")
plt.title("Displacement of Concorde 102 Main Gear")
plt.legend()
plt.grid(True)
plt.show()

# Plotting RK4 Velocity
plt.figure(figsize=(12, 6))
for h in h_values:
    t_r, y_r, v_r = rk4_method(h, m, c, k, t0, tf, y0, v0)
    
    plt.plot(t_r, v_r, label=f"RK4, h={h}")
plt.xlabel("Time (s)")
plt.ylabel("Velocity v(t) (m/s)")
plt.title("Velocity of Concorde 102 Main Gear Response")
plt.legend()
plt.grid(True)
plt.show()

# Phase portrait of rk4
plt.figure(figsize=(12, 6))
for h in h_values:
    t_r, y_r, v_r = rk4_method(h, m, c, k, t0, tf, y0, v0)
    
    plt.plot(y_r, v_r, label=f"RK4, h={h}")
plt.xlabel("Displacement")
plt.ylabel("Velocity")
plt.title("Velocity vs Displacement using RK4")
plt.legend()
plt.grid(True)
plt.show()

# Plot RK4 error
plt.figure(figsize=(12, 6))

for h in h_values:
    t_r, y_r, v_r = rk4_method(h, m, c, k, t0, tf, y0, v0)
    error_r = calculate_error(t_r, y_r, m, c, k, v0)
    
    plt.plot(t_r, error_r, label=f"RK4 h={h}")
plt.xlabel("Time (s)")
plt.ylabel("Absolute Error (m)")
plt.title("RK4 Method Error Over Time")
plt.legend()
plt.grid(True)
plt.show()

# Log-log plot
plt.figure(figsize=(10, 6))
plt.loglog(h_values, euler_max_errors, label="Euler")
plt.loglog(h_values, rk4_max_errors, label="RK4")
plt.xlabel("Step size h")
plt.ylabel("Max Absolute Error")
plt.title("Log-Log Plot: Error vs Step Size")
plt.legend()
plt.grid(True)
plt.show()
```
