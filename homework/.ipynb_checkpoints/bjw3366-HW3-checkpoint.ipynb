```python

import numpy as np
import matplotlib.pyplot as plt
import math

#hw2 got canceled, look at hw3
```

```python
def sinTaylor(x,numSamples):
    sinEst = 0
    for i in range(numSamples):
        numerator = ((-1)**i) * (x **(2 * i + 1))
        denominator = math.factorial(2 * i + 1)
        sinEst += numerator / denominator
    return sinEst
        
```

```python
def main():
    numSamples = np.array([1,2,3,4])
    x = np.arange(0,10.1,0.1)
    sinx = np.empty(len(x))
    sinEstOneTerm = np.empty(len(x))
    sinEstTwoTerm = np.empty(len(x))
    sinEstThreeTerm = np.empty(len(x))
    sinEstFourTerm = np.empty(len(x))
    
    for i in range(len(x)):
        sinEstOneTerm[i] = sinTaylor(x[i],numSamples[0])
        sinEstTwoTerm[i] = sinTaylor(x[i],numSamples[1])
        sinEstThreeTerm[i] = sinTaylor(x[i],numSamples[2])
        sinEstFourTerm[i] = sinTaylor(x[i],numSamples[3])
        sinx[i] = np.sin(x[i])
        
    absErrorOne = np.abs(sinx - sinEstOneTerm)
    absErrorTwo = np.abs(sinx - sinEstTwoTerm)
    absErrorThree = np.abs(sinx - sinEstThreeTerm)
    absErrorFour = np.abs(sinx - sinEstFourTerm)
    
    plt.figure()
    plt.plot(x,absErrorOne, label = "1 Term")
    plt.plot(x,absErrorTwo, label = "2 Terms")
    plt.plot(x,absErrorThree, label = "3 Terms")
    plt.plot(x,absErrorFour, label = "4 Terms")
    plt.xlabel("X Values")
    plt.ylabel("Error of sin(x)")
    plt.title("Absolute Error of Sin(x)")
    plt.legend()
    plt.show()
    
    
    plt.figure()
    plt.plot(x,absErrorOne)
    plt.xlabel("X Values")
    plt.ylabel("Error of sin(x)")
    plt.title("Absolute Error of Sin(x) with One Term")
    plt.show()
    
    plt.figure()
    plt.plot(x,absErrorTwo)
    plt.xlabel("X Values")
    plt.ylabel("Error of sin(x)")
    plt.title("Absolute Error of Sin(x) with Two Terms")
    plt.show()
    
    plt.figure()
    plt.plot(x,absErrorThree)
    plt.xlabel("X Values")
    plt.ylabel("Error of sin(x)")
    plt.title("Absolute Error of Sin(x) with Three Terms")
    plt.show()
    
    plt.figure()
    plt.plot(x,absErrorFour)
    plt.xlabel("X Values")
    plt.ylabel("Error of sin(x)")
    plt.title("Absolute Error of Sin(x) with Four Terms")
    plt.show()
    
    
    return 0
if __name__ == "__main__":
    main()
```

```python
# The more terms that are added to the taylor series, more x values are accurate.
# For example, plot 1 is only accurate to about x = 0.5
# where as plot 4 is accuratae until x < 6
# The error does not decrease the same for all x. For small x, the error is already very small with few terms, 
# but for large x, more terms are needed to get good accuracy.
```
