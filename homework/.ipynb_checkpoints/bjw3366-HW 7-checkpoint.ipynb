```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x, y):
     # Function representing dy/dx = -2ycos(x)
    return -2 * y * np.cos(x)
```

```python
# Step Size
h_values = np.arange(0.05, 1.05, 0.05)
h_values = np.insert(h_values, 0, 0.01)

# Initalizing arrays to use later
avg_list = np.array([])
error_list = np.array([])
local_error_list = np.array([])
y_values_15 = np.array([])

for h in h_values:
    # Define parameters
    x_start = 0
    x_end = 10 * np.pi
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    
    # Initial condition
    y_values[0] = 1  # y(0) = 1
    
    # Implement Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])
    
    # Average y value for each h
    y_avg = np.mean(y_values)
    avg_list = np.append(avg_list, y_avg)
    
    # Avg Error
    y_exact = np.exp(-2 * np.sin(x_values))   # The function y' = -2ycos(x) solved for y
    error = np.abs(y_values - y_exact)
    error_list = np.append(error_list, np.mean(error))
    
    # Local Error
    distance_from_15 = np.abs(x_values - 15)
    x_15 = np.argmin(distance_from_15)
    
    y_values_15 = np.append(y_values_15, y_values[x_15])
    
    local_error = np.abs(y_values[x_15] - y_exact[x_15])
    local_error_list = np.append(local_error_list, local_error)
    
    plt.plot(x_values, y_values)
plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2ycos(x) using Euler's Method")
plt.show()

plt.plot(h_values, error_list)
plt.xlabel('h')
plt.ylabel('Avg Error')
plt.title("Avg Error at Each Step")
plt.show()

plt.plot(h_values, local_error_list)
plt.xlabel('h')
plt.ylabel('Local Error')
plt.title("Local Error at Each Step when x ~ 15")
plt.show()

print("Y values for each run of h")
print(avg_list)
print("\nAvg error for each run of h")
print(error_list)
print("\nThe y values when x ~ 15 for each run of h")
print(y_values_15)
print("\nLocal error at x ~ 15 for each run of h")
print(local_error_list)
```
