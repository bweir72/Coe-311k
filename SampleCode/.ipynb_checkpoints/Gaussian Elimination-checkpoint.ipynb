# Gaussian Elimination

```python
# Solving a matrix by rearanging the elements to make all the elements below the main diagonal 0

import numpy as np
```

```python
def gaussian_elimination(A, b):    # A is matrix; b is constants, what the equations =
    n = len(b)
    # Forward elimination
    for i in range(n):
        # Pivoting
        max_row = max(range(i, n), key=lambda x: abs(A[x][i]))
        A[i], A[max_row] = A[max_row], A[i]
        b[i], b[max_row] = b[max_row], b[i]
        
        # Make the rows below this one 0 in current column
        for j in range(i + 1, n):
            ratio = A[j][i] / A[i][i]
            A[j] = [A[j][k] - ratio * A[i][k] for k in range(n)]
            b[j] -= ratio * b[i]
    
    # Backward substitution
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - sum(A[i][j] * x[j] for j in range(i + 1, n))) / A[i][i]
        
    return x
```

```python
# Example Usage
A = [[2, 3, -1], [4, -2, 2], [-2, 1, 2]]
b = [5, 6, -1]
solution = gaussian_elimination(A, b)
print(f"Solution: {solution}")
```

```python
ans = np.linalg.solve(A,b)
print(ans)
```
