```python
import numpy as np
```

```python
v1 = np.array([1,2,3])
v2 = np.array([4,5,6])
print("Vector 1:", v1)
print("Vector 2:", v2)
```

```python
v_add = v1 + v2
v_sub = v1 - v2
print(v_add)
print(v_sub)
```

```python
v_scaled = v2 * 2
print(v_scaled)
```

```python
v1_dot_v2 = np.dot(v1,v2)
print("The dot product of v1 and v2 is", v1_dot_v2)
```

# The norm of a vector

```python
v1_norm = np.linalg.norm(v1)
print("The norm of v1 is", v1_norm)
```

```python
#Creating matices
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[1,1,1],[1,1,1],[1,1,1]])
C = np.array([[4,2,6],[5,2,1]])
print(A)
print("\n",B)


matrixAdd = A + C
print("\n",matrixAdd)

```

# Tranposing a Matrix

```python
transpose_C = C.T
print(transpose_C)
D = ([[1,2],[3,6],[8,5]])

# ixj * gxh only works if j and g are the same and the result will be ixh matrix
matrix_Multiplication = np.dot(D,C)
print("\n",matrix_Multiplication)
```

# Solving System of Linear Equations

```python
# Ax = b
# A is the coefficient matrix
# x is the unkown matrix
# b is the constant matrix

A1 = np.array([[2, 3], [4, -1]])
b = np.array([5, 3])
x = np.linalg.solve(A1,b)
print(x)
```

# Inverse of a Matrix

```python
# The inverse of a matrix 𝐴, denoted as 𝐴−1 , is the matrix that, when
# multiplied by 𝐴 results in the identity matrix:
# A * A^-1 = I
# The identity matrix is every is 0 except the diagonal that is 1s

A3 = np.array([[4,7],[2,6]])
A_inverse = np.linalg.inv(A)
print(A_inverse)

# Check to make sure
I = np.dot(A,A_inverse)
print(I)
```
