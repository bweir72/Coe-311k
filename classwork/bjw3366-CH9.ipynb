```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Part 1

# Data points
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix with a column of ones (for the intercept)
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Slope: {m}, Intercept: {c}")

y1 = m * X + c

plt.figure()
plt.scatter(X,Y)
plt.plot(X,y1)
plt.title("Original Best Fit Line")
plt.show()
```

```python
# Part 2

# Adding noise
X2 = X + np.random.rand(len(X2))
Y2 = Y + np.random.rand(len(Y2))

# Stack the design matrix with a column of ones (for the intercept)
A2 = np.vstack([X2, np.ones(len(X2))]).T

# Solve using least squares
m2, c2 = np.linalg.lstsq(A2, Y2, rcond=None)[0]
print(f"Slope: {m2}, Intercept: {c2}")

y2 = m2 * X2 + c2

plt.figure()
plt.scatter(X2,Y2)
plt.plot(X2,y2)
plt.title("Best Fit Line w/ Noise")
plt.show()
```
