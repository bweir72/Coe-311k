# ICE Leibniz Series for 𝜋

```python
import numpy as np
```

```python
piEst = np.empty(5000)
relError = np.empty(50)

for i in range(5000):
    numerator = (-1)**i
    denomerator = 2 * i + 1
    piEst[i] =  4 * numerator / denomerator
    
for i in range(0, 5000, 100):
    relError = abs((np.pi - piEst[0:i]) / np.pi)
    
```
