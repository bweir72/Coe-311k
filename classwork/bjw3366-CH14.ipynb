```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the function and the domain
f = lambda x: np.cos(x)
a, b = 0, 2 * np.pi # Interval
N = 50 # Number of sample points

# Create a uniform grid
x = np.linspace(a, b, N)
y = f(x)

h = (b - a) / (N -1) # Step Size

# Central Difference
df_dx_central = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference
df_dx_forward = (f(x + h) - f(x)) / h

# Backward Difference
df_dx_backward = (f(x) - f(x -h)) / h

# True derivative of cos(x)
true_derivative = -np.sin(x)

plt.figure()
plt.plot(x, y, label = 'cos(x)')
plt.plot(x, df_dx_central, label = 'Central Difference')
plt.plot(x, df_dx_forward, label = 'Forward Difference')
plt.plot(x, df_dx_backward, label = 'Backward Difference')
plt.plot(x, true_derivative, label = 'Derivative of cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()

```
