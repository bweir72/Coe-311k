```python
import numpy as np
import matplotlib.pyplot as plt
import random as rng
```

```python
def appxPi(numSamples):
    counter = 0
    for i in range (numSamples):
        x = rng.uniform(-1,1)
        y = rng.uniform(-1,1)
        if( x**2 + y**2 < 1):
            counter += 1
    
    pi = 4 * counter / numSamples
    return pi
```

```python
def main():
    numSamples = 1000
    runs = 100
    piEst = np.array([])
    x = np.arange(1,101)
    
    for i in range(runs):
        piEst = np.append(piEst, appxPi(numSamples))
    
    trueError = np.abs(np.pi - piEst)
    relativeError = (np.abs(np.pi - piEst)) / np.pi
    
    plt.figure()
    plt.plot(x,trueError)
    plt.xlabel("Runs")
    plt.ylabel("True Error")
    plt.title("True Error of Estimating Pi")
    plt.show()
    
    plt.figure()
    plt.plot(x,relativeError)
    plt.xlabel("Runs")
    plt.ylabel("Relative Error")
    plt.title("Relative Error of Estimating Pi")
    plt.show()
    
    
if __name__ == "__main__":
    main()
```
