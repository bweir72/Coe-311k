```python
import numpy as np
import time
import matplotlib.pyplot as plt
```

```python
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
```

```python
def main():
    
    # 3x3
    M1a = np.random.randint(1, 10, (3, 3))
    M1b = np.random.randint(1, 10, (3, 3))

    # 6x6
    M2a = np.random.randint(1, 10, (6, 6))
    M2b = np.random.randint(1, 10, (6, 6))

    # 12x12
    M3a = np.random.randint(1, 10, (12, 12))
    M3b = np.random.randint(1, 10, (12, 12))

    # 24x24
    M4a = np.random.randint(1, 10, (24, 24))
    M4b = np.random.randint(1, 10, (24, 24))

    # 48x48
    M5a = np.random.randint(1, 10, (48, 48))
    M5b = np.random.randint(1, 10, (48, 48))

    # 96x96
    M6a = np.random.randint(1, 10, (96, 96))
    M6b = np.random.randint(1, 10, (96, 96))

    # 192x192
    M7a = np.random.randint(1, 10, (192, 192))
    M7b = np.random.randint(1, 10, (192, 192))

    # 384x384
    M8a = np.random.randint(1, 10, (384, 384))
    M8b = np.random.randint(1, 10, (384, 384))

    matricesA = [M1a, M2a, M3a, M4a, M5a, M6a, M7a, M8a]
    matricesB = [M1b, M2b, M3b, M4b, M5b, M6b, M7b, M8b]

    naive_times = []
    numpy_times = []
    matrix_nums = [1, 2, 3, 4, 5, 6, 7, 8]
    
    for i in range(8):
        A = matricesA[i]
        B = matricesB[i]

        # naive timing
        start = time.perf_counter()
        C_naive = naive_matrix_multiplication(A, B)
        end = time.perf_counter()
        naive_times.append(end - start)

        # numpy timing
        start = time.perf_counter()
        C_numpy = np.dot(A, B)
        end = time.perf_counter()
        numpy_times.append(end - start)

    plt.plot(matrix_nums, naive_times, marker='o', label='Naive')
    plt.plot(matrix_nums, numpy_times, marker='o', label='NumPy')
    plt.xlabel("Matrix Number")
    plt.ylabel("Runtime (seconds)")
    plt.title("Naive vs NumPy Matrix Multiplication")
    plt.xticks(matrix_nums)
    plt.legend()
    plt.grid(True)
    plt.show()
    
    
    return 0

if __name__ == "__main__":
    main()
```
